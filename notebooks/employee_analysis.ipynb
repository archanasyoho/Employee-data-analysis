```python
#Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Load dataset
df = pd.read_csv(r"C:\Users\Admin\Desktop\Training\Python Training\Employee_data_analysis\data\IBM-HR-Employee-Attrition.csv")

print("Dataset loaded successfully!")
```

```python
#Check dataset information
df.info()
```

```python
#Statistical summary
df.describe()
```

```python
#Check missing values
print("Missing Values:")
print(df.isna().sum())
```

```python
#Check duplicate records
print("Duplicate Records:")
print(df.duplicated().sum())

```

```python
#Check column names
print("Dataset Columns:")
print(df.columns.tolist())

```

```python
#Employee Attrition Analysis
print("Employee Attrition Count:")
print(df["Attrition"].value_counts())
```

```python
#Department-wise Attrition
department_attrition = pd.crosstab(
    df["Department"],
    df["Attrition"]
)
print(department_attrition)

```

```python
#Attrition Percentage
attrition_percentage = (
    df["Attrition"].value_counts(normalize=True) * 100
)
print(attrition_percentage)
```

```python
#Department-wise Attrition Chart
sns.countplot(
    data=df,
    x="Department",
    hue="Attrition"
)
plt.title("Department-wise Employee Attrition")

plt.xlabel("Department")

plt.ylabel("Number of Employees")

plt.xticks(rotation=15)

plt.show()

```

```python
#Create Data Dictionary
data_dictionary = pd.DataFrame({
    "Column Name": df.columns,
    "Data Type": df.dtypes.astype(str).values,
    "Non-Null Count": df.notna().sum().values,
    "Missing Values": df.isna().sum().values
})

data_dictionary


```

```python
#Save Data Dictionary
data_dictionary.to_csv(r"C:\Users\Admin\Desktop\Training\Python Training\Employee_data_analysis\reports\data_dictionary.csv",
    index=False)

print("Data Dictionary saved successfully!")
```
